### PET - FDG only baseline performance analysis

```python
import seaborn as sn
import pandas as pd
import matplotlib.pyplot as plt
import os
```

```python
exp_path = "../experiments/pet_fdg_baseline/20260301_171206"
```

```python
cm_sum = None
report_sum = None
# Average confusion matrix
for fold in range(5):
    
    fold_path = os.path.join(exp_path, "fold_" + str(fold))

    cm = pd.read_csv(os.path.join(fold_path, "confusion_matrix.csv"), index_col=0)
    report = pd.read_csv(os.path.join(fold_path, "classification_report.csv"), index_col=0)

    if fold == 0:
        cm_sum = cm
        report_sum = report
    else:
        cm_sum += cm
        report_sum += report

mri_cm_avg = cm_sum / cm_sum.sum().sum()
report_avg = report_sum/5

lbls = ["CN", "MCI", "AD"]
mri_cm_avg.index = lbls
mri_cm_avg.columns = lbls

cmap = sn.color_palette('Blues', as_cmap=True)
sn.heatmap(mri_cm_avg, annot=True, cmap=cmap)

print(report_avg)
```
