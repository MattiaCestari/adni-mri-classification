```python
import pandas as pd
```

```python
df_mri = pd.read_csv("../csv/oasis/oasis3_mri_all.csv")

# Loop through experiments and keep only those with T1w
df_mri = df_mri[ df_mri["Scans"].notna() ]
df_mri = df_mri[ df_mri["Scans"].str.contains("T1w") ]
df_mri = df_mri[ df_mri["Scanner"].notna() ]

# Rename columns
df_mri = df_mri.rename(columns={"MR ID":"image_id", "Subject":"subject_id"} )

# Add group column
df_mri["group"] = "MRI-T1-3T"
df_mri.loc[ list( df_mri["Scanner"] == "1.5T" ), "group"] = "MRI-T1-1.5T"

# Keep only relevant columns
df_mri = df_mri[["image_id", "subject_id", "group"]]
```

```python
df_pet = pd.read_csv("../csv/oasis/oasis3_pet_all.csv")
df_pet = df_pet.rename(columns={"XNAT_PETSESSIONDATA ID":"image_id", "Subject":"subject_id"})

df_pet["group"] = df_pet["image_id"].apply( lambda s : "PET-" + s.split("_")[1])
df_pet = df_pet[ [ "image_id", "subject_id", "group"]]
```

```python
df_final = pd.concat([df_mri, df_pet])
df_final.to_csv("../csv/oasis/dataset_multimodal.csv")
```
